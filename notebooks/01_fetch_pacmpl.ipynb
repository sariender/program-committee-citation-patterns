# PACMPL Data Extraction

This notebook fetches all works published under ISSN 2475-1421 [(Proceedings of the ACM on Programming Languages)](https://dl.acm.org/loi/pacmpl/group/d2010.y2017) from [OpenAlex](https://openalex.org/) and stores raw JSON for **reproducibility.**
[pyalex](https://github.com/J535D165/pyalex) is the library that I use.

```python
from pyalex import Works
from itertools import chain
import json
from pathlib import Path
```

```python
ISSN = "2475-1421" # PACMPL
DATA_DIR = Path("../data/raw")
OUTPUT_FILE = DATA_DIR / "pacmpl_full.jsonl"
```

```python
query = Works().filter(primary_location={"source": {"issn": ISSN}})
results = []
for w in chain(*query.paginate(per_page=200,n_max=None)):
    results.append(w)
print("Total works fetched:", len(results))
```

```python
total_api = query.count()
print("Expected:", total_api)
print("Fetched:", len(results))
```

```python
with open(OUTPUT_FILE, "w") as f:
    for row in results:
        f.write(json.dumps(row) + "\n")

print("Saved to:", OUTPUT_FILE)
```

```python
years = sorted(set(w["publication_year"] for w in results))
print(years)
```
